# **Qué es el análisis de sentimiento**

Es una tarea de clasificación de texto. Dado un texto, el modelo decide si la intención utilizada es positiva, negativa o neutral.

No es "**entender**" el texto como lo haría una persona. Es detectar patrones de palabras y contexto que, estadísticamente, se asocian a cada categoría.

Para esta práctica en español, el modelo más adecuado es **RoBERTuito**. Es un modelo entrenado específicamente con **500 millones de tweets en español**, lo que le permite entender jerga, emojis, abreviaturas y texto informal.

Está diseñado para redes sociales, que es donde el análisis de sentimiento tiene más aplicación real. Y funciona bien en tareas de sentimiento, odio e ironía en español.

RoBERTuito no se entrenó desde cero. Es un modelo preentrenado: alguien invirtió muchísimo tiempo y recursos en entrenarlo con textos masivos para que "entienda" el español.

Luego ese modelo base se ajustó (mediante fine-tuning) con un conjunto más pequeño de tweets etiquetados como positivos, negativos o neutros. El resultado es un modelo especializado en análisis de sentimiento en español, listo para usar.

Esto es importante porque explica por qué tú no entrenas nada. Usas un modelo que ya existe, ya está entrenado, y solo lo llamas desde Colab. Es exactamente lo que haría un desarrollador profesional: **no reinventar la rueda, usar lo que ya funciona**.

# **Hugging Face**

Hugging Face es la plataforma donde viven estos modelos. Su función **pipeline** es una abstracción: tú le dices "quiero realizar un análisis de sentimiento" y "usa este modelo", y ella se encarga de todo lo demás (preprocesar el texto, pasarlo por el modelo, devolver el resultado).

Solo se necesita entender que pipeline es el "botón" que conecta el texto con un modelo ya entrenado.

El modelo devuelve dos cosas:

    Una etiqueta: POS, NEG o NEU

    Una confianza (score), un número entre 0 y 1

La confianza no es lo mismo que **"acierto"**. Es la seguridad del modelo en su propia predicción. Un score de 0.99 no garantiza que tenga razón; un score de 0.51 en una frase ambigua es normal y no es un fallo.

El modelo no lee intenciones, lee patrones. El guiño o el signo de interrogación alteran el contexto que el modelo aprendió durante el preentrenamiento, y eso mueve la predicción.

```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis", model="pysentimiento/robertuito-sentiment-analysis")
# carga un modelo de inteligencia artificial
# entrenado para analizar el sentimiento
# de textos en español, con especial enfoque
# en lenguaje informal y redes sociales

# === este es el texto  que se clasifica ===
texto = "Me encanta este módulo de IA"
# =====================================

resultado = clasificador(texto)
print(resultado)
```

```python
# clasifica un texto que sea negativo

texto = ""
resultado = clasificador(texto)
print(resultado)
```

```python
# clasifica un texto neutro
texto = "Genial, tardaron un mes entero en responder a un simple correo. ¡Un aplauso!"
resultado = clasificador(texto)
print(texto)
print(resultado)
```

<h1>¿Cómo funciona?</h1>
El análisis de sentimiento es una tarea de clasificación automática de texto. Funciona mediante modelos de inteligencia artificial (como RoBERTuito) que han sido entrenados con millones de textos para detectar patrones estadísticos de palabras, contexto, emojis y jerga. Cuando le pasas un texto, el modelo evalúa esos patrones y emite una etiqueta (positivo, negativo o neutral) acompañada de un score o confianza (un número entre 0 y 1 que indica qué tan seguro está de su predicción).

<h1>¿Qué límites tiene?</h1>
No comprende el significado real: No "entiende" como lo haría una persona, sino que busca correlaciones estadísticas aprendidas durante su entrenamiento.

Dificultad con la ironía y el sarcasmo: Al basarse en palabras y patrones, un comentario irónico ("¡Qué gran servicio, tardaron un mes en responder!") puede ser malinterpretado fácilmente como positivo si contiene palabras elogiosas fuera de contexto.

La ambigüedad: Frases ambiguas o con doble sentido confunden al modelo, lo que se refleja en scores cercanos a 0.5 (baja confianza).

El score no es un certificado de acierto: Una confianza de 0.99 significa que el modelo está muy seguro según sus patrones, pero no garantiza al 100% que la clasificación sea objetivamente correcta.


# **Práctica**

Implementa en python un analizador de comentarios. Los comentarios positivos y nuetros serán aceptados, salvo que contengan alguna palabra de la lista negra de palabras. Los comentarios negativos que contengan alguna palabra de la lista negra o que tengan un score de más de 0.5 serán rechazadas.

```python
# pon aquí el código de tu analizador

from transformers import pipeline

# 1. Cargar el modelo de análisis de sentimientos en español (RoBERTuito)
clasificador = pipeline("sentiment-analysis", model="pysentimiento/robertuito-sentiment-analysis")

# 2. Definir la lista negra de palabras prohibidas
lista_negra = ["PSOE", "odio", "basura", "rata", "horrible", "imbecil"]

def analizar_comentario(texto):
    # Obtener el resultado del modelo
    resultado = clasificador(texto)[0]
    label = resultado['label']   # 'POS', 'NEU' o 'NEG'
    score = resultado['score']   # Confianza del modelo (entre 0 y 1)

    # Comprobar si el texto contiene alguna palabra de la lista negra
    texto_lower = texto.lower()
    contiene_palabra_prohibida = any(palabra in texto_lower for palabra in lista_negra)

    # --- REGLAS DE NEGOCIO ---

    # Comentarios positivos y neutros:
    # Aceptados, salvo que contengan alguna palabra de la lista negra.
    if label in ["POS", "NEU"]:
        if contiene_palabra_prohibida:
            return f"RECHAZADO  [Etiqueta: {label}] (Contiene palabra de la lista negra. Score: {score:.2f})"
        else:
            return f"ACEPTADO  [Etiqueta: {label}] (Score: {score:.2f})"

    # Comentarios negativos:
    # Rechazados si contienen alguna palabra de la lista negra O score > 0.5.
    elif label == "NEG":
        if contiene_palabra_prohibida or score > 0.5:
            return f"RECHAZADO  [Etiqueta: {label}] (Score > 0.5 o palabra de lista negra. Score: {score:.2f})"
        else:
            return f"ACEPTADO  [Etiqueta: {label}] (Negativo con score <= 0.5 y sin palabras prohibidas. Score: {score:.2f})"

    return "RECHAZADO"

# 3. Lista de 10 frases variadas
frases_prueba = [
    # --- 2 Frases Positivas ---
    "El curso está muy bien explicativo y he aprendido bastante.",
    "Me encanta este producto, funciona genial y la calidad es excelente.",

    # --- 2 Frases Negativas ---
    "Como votes al PSOE te voy a pegar",
    "Eres un pedazo de rata, he perdido mi dinero.",

    # --- 2 Frases Neutras ---
    "El paquete fue entregado hoy a las tres de la tarde, imbecil.",
    "El libro consta de diez capítulos e incluye un índice al final.",

    # --- 2 Frases Irónicas ---
    "Qué maravilla, esperé dos horas en la fila para que me digan que vuelva mañana.",
    "Genial, la pantalla se rompió a los diez minutos de encenderlo por primera vez.",

    # --- 2 Frases con Emoticonos ---
    "¡Qué gran día para disfrutar con amigos! 😊🎉",
    "No me ha gustado nada la comida de este sitio 🤮👎"
]

# 4. Procesar e imprimir los resultados
print("--- RESULTADOS DEL ANALIZADOR DE COMENTARIOS ---\n")
for i, comentario in enumerate(frases_prueba, 1):
    estado = analizar_comentario(comentario)
    print(f"{i}. Texto: '{comentario}'")
    print(f"   Resultado: {estado}\n")
```
